This notebook provides a clean, end-to-end baseline for sentiment classification using IMDb movie descriptions and Hugging Face Transformers.

```python
import os
import re
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, classification_report, confusion_matrix

from datasets import Dataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification, TrainingArguments, Trainer, DataCollatorWithPadding

import transformers, accelerate
print('transformers:', transformers.__version__)
print('accelerate:', accelerate.__version__)
# Runtime profile for slower computers
FAST_MODE = True  # Set False for final full-quality run
SEED = 42
np.random.seed(SEED)
```

## Load and Prepare Data

```python
df = pd.read_csv('Dataset/IMDb movies.csv', low_memory=False)
df = df.dropna(subset=['description', 'avg_vote']).copy()

# Create binary sentiment labels from rating
df['sentiment'] = np.where(df['avg_vote'] >= 6, 'positive', 'negative')
df['label'] = df['sentiment'].map({'negative': 0, 'positive': 1}).astype(int)

def clean_text(x):
    x = str(x).lower()
    x = re.sub(r'<.*?>', ' ', x)
    x = re.sub(r'http\\S+|www\\S+', ' ', x)
    x = re.sub(r'[^a-z\\s]', ' ', x)
    x = re.sub(r'\\s+', ' ', x).strip()
    return x

df['text'] = df['description'].apply(clean_text)
df = df[df['text'].str.len() > 0].copy()

# Runtime-aware sampling: faster in FAST_MODE
max_samples = 6000 if FAST_MODE else 20000
if len(df) > max_samples:
    df = df.sample(n=max_samples, random_state=SEED).reset_index(drop=True)

print('FAST_MODE:', FAST_MODE)
print('Shape:', df.shape)
print(df['sentiment'].value_counts())
```

```python
train_texts, temp_texts, train_labels, temp_labels = train_test_split(
    df['text'],
    df['label'],
    test_size=0.3,
    random_state=42,
    stratify=df['label']
)

val_texts, test_texts, val_labels, test_labels = train_test_split(
    temp_texts,
    temp_labels,
    test_size=0.5,
    random_state=42,
    stratify=temp_labels
)

print('Train size:', len(train_texts))
print('Val size:', len(val_texts))
print('Test size:', len(test_texts))
```

## Tokenization and Dataset Creation

```python
cache_dir = './hf_cache'
os.makedirs(cache_dir, exist_ok=True)

# Assignment requirement: bert-base-uncased
model_name = 'bert-base-uncased'
tokenizer = AutoTokenizer.from_pretrained(model_name, cache_dir=cache_dir)

MAX_LEN = 32 if FAST_MODE else 48

def tokenize_texts(texts):
    # Dynamic padding + shorter max_length keeps memory and runtime manageable.
    return tokenizer(list(texts), truncation=True, padding=False, max_length=MAX_LEN)

train_enc = tokenize_texts(train_texts)
val_enc = tokenize_texts(val_texts)
test_enc = tokenize_texts(test_texts)

train_ds = Dataset.from_dict({
    'input_ids': train_enc['input_ids'],
    'attention_mask': train_enc['attention_mask'],
    'labels': list(train_labels)
})

val_ds = Dataset.from_dict({
    'input_ids': val_enc['input_ids'],
    'attention_mask': val_enc['attention_mask'],
    'labels': list(val_labels)
})

test_ds = Dataset.from_dict({
    'input_ids': test_enc['input_ids'],
    'attention_mask': test_enc['attention_mask'],
    'labels': list(test_labels)
})

print('MAX_LEN:', MAX_LEN)
```

## Model, Fine-Tuning, and Experiments
We run three setups for clear comparison:
1. Full BERT fine-tuning (baseline) 
2. Freeze all BERT encoder layers and train classifier only 
3. Fine-tune only the last 2 BERT encoder layers plus classifier


```python
import inspect
import torch

data_collator = DataCollatorWithPadding(tokenizer=tokenizer)
arg_names = set(inspect.signature(TrainingArguments.__init__).parameters.keys())
eval_arg_name = 'evaluation_strategy' if 'evaluation_strategy' in arg_names else 'eval_strategy'

BATCH_SIZE = 1
GRAD_ACCUM = 2 if FAST_MODE else 4
NUM_EPOCHS = 1
MAX_STEPS = 300 if FAST_MODE else -1

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=-1)
    p, r, f1, _ = precision_recall_fscore_support(labels, preds, average='binary', zero_division=0)
    acc = accuracy_score(labels, preds)
    return {'accuracy': acc, 'precision': p, 'recall': r, 'f1': f1}

def set_trainable_layers(model, mode='full'):
    for p in model.parameters():
        p.requires_grad = False

    if mode == 'full':
        for p in model.parameters():
            p.requires_grad = True
    elif mode == 'freeze_all':
        for p in model.classifier.parameters():
            p.requires_grad = True
    elif mode == 'last2':
        for p in model.bert.encoder.layer[-2:].parameters():
            p.requires_grad = True
        for p in model.classifier.parameters():
            p.requires_grad = True
    else:
        raise ValueError('mode must be one of: full, freeze_all, last2')

def run_experiment(exp_name, mode):
    model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=2, cache_dir=cache_dir)
    set_trainable_layers(model, mode=mode)

    ta_kwargs = dict(
        output_dir=f'./results/{exp_name}',
        save_strategy='no' if FAST_MODE else 'epoch',
        learning_rate=2e-5,
        per_device_train_batch_size=BATCH_SIZE,
        per_device_eval_batch_size=BATCH_SIZE,
        gradient_accumulation_steps=GRAD_ACCUM,
        num_train_epochs=NUM_EPOCHS,
        max_steps=MAX_STEPS,
        weight_decay=0.01,
        logging_dir=f'./logs/{exp_name}',
        logging_steps=50,
        dataloader_num_workers=0,
        eval_accumulation_steps=8,
        optim='adamw_torch',
        report_to='none'
    )
    ta_kwargs[eval_arg_name] = 'no' if FAST_MODE else 'epoch'
    training_args = TrainingArguments(**ta_kwargs)

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=train_ds,
        eval_dataset=val_ds,
        compute_metrics=compute_metrics,
        data_collator=data_collator
    )

    trainer.train()
    return trainer

experiments = [('full_finetune', 'full'), ('freeze_encoder', 'freeze_all'), ('finetune_last2', 'last2')]

print('FAST_MODE:', FAST_MODE, '| max_samples:', len(df), '| max_len:', MAX_LEN, '| max_steps:', MAX_STEPS)
trained = {}
for exp_name, mode in experiments:
    print(f'\\nRunning: {exp_name} ({mode})')
    trained[exp_name] = run_experiment(exp_name, mode)
```

```python
comparison_rows = []
conf_mats = {}

for exp_name, _ in experiments:
    trainer = trained[exp_name]
    metrics = trainer.evaluate(test_ds)
    pred = trainer.predict(test_ds)

    y_pred = np.argmax(pred.predictions, axis=1)
    y_true = np.array(test_labels)

    cm = confusion_matrix(y_true, y_pred)
    conf_mats[exp_name] = cm

    comparison_rows.append({
        'experiment': exp_name,
        'accuracy': metrics.get('eval_accuracy'),
        'precision': metrics.get('eval_precision'),
        'recall': metrics.get('eval_recall'),
        'f1': metrics.get('eval_f1')
    })

    print(f'\\n=== {exp_name} ===')
    print(metrics)
    print('Confusion Matrix:')
    print(cm)
    print('Classification Report:')
    print(classification_report(y_true, y_pred, zero_division=0))

comparison_df = pd.DataFrame(comparison_rows).sort_values(by='f1', ascending=False).reset_index(drop=True)
print('\\nExperiment Comparison (sorted by F1):')
display(comparison_df)
```
